## 반복문

### for 문

```python
# 반복문 연습1
for i in range(10):
    print(i)
```

```python
# 반복문 연습2
s = 0
for i in range(1, 101):
    s = s + i
print("1 + 2 + ... + 100 =", s)
```

```python
# 반복문 연습3
list_var = [i for i in range(1,5+1)]
for i in list_var:
    j = i*10
    print(i,j)
```

```python
# 반복문 연습4
dict_var = {'key1': 10, 'key2': 20, 'key3': 30}
for j in dict_var:
    print(j, dict_var[j])
```

```python
# 반복문 연습5
fruits = ["peach", "water melon", "apple", "lemon", "pear"]
for i, fruit in enumerate(fruits):
    print(f"{i+1}번 과일은 {fruit}입니다.")
```

```python
# 반복문 연습6
for i in range(2, 10):
    print(f"=== {i}단 ===") # 제목 출력
    for j in range(1, 10):
        print(f"{i}x{j}={i*j}")
    print() # 빈줄 출력
```

```python
# 반복문 연습7
nums = [i for i in range(1,11)]
for k in nums:
    if k%3 == 0:
        continue
    print(k)
    if k > 7:
        break
```

### while 문

```python
# 반복문 연습8
s = 0
i = 0
while i < 100:
    i = i + 1
    s = s + i
print("1 + 2 + ... + 100 =", s)
```

```python
# 반복문 연습9
epsilon = 1.0
decay_factor = 0.9
i = 1
while True:
    epsilon = epsilon*decay_factor
    if epsilon < 1e-7:
        break
    i = i + 1
print(i, epsilon)
```

## 함수

```python
# 입력값, 반환값 모두 없는 함수
def greeting():
    print("How are you dong?")

# 반환값이 없는 함수
def even_odd(a):
    if a%2 == 0:
        print("Even number")
    else:
        print("Odd number")

# 입력값, 반환값 모두 있는 함수
def square_sum(a,b):
    return a**2 + b**2
```

```python
# 함수 연습2
greeting()
even_odd(10)
result = square_sum(3,4)
print(result)
```

```python
# 함수 연습3
def bonus(a, b, c=10):
    res = a + b + c
    return res

res1 = bonus(10, 20)
res2 = bonus(10, 20, c=15)
print(res1, res2)
```

```python
# 함수 연습4
a = 10
def my_func(x):
    y = 20
    res = x + y + a
    return res
res = my_func(3)
print(res)
```

```python
# 함수 연습5
print(f"a={a}")
print(f"y={y}")
```

```python
# 함수 연습6
a = 10
def my_func(x):
    global a
    global y
    y = 20
    a = a + 1
    res = x + y + a
    return res
res = my_func(3)
print(res)
print(f"y={y}")
```
