# Numpy

```python
# 넘파이 불러오기
import numpy as np
```

```python
# 리스트를 사용한 1차원 배열의 생성
list_var1 = [1,2,3,4]
arr1 = np.array([1,2,3,4])
print(list_var1, type(list_var1))
print(arr1, type(arr1))
print(arr1.shape, len(arr1))
```

```python
# 스칼라 곱하기
print(3*list_var1)
print(3*arr1)
```

```python
# 배열에 스칼라 더하기
print(arr1 + 10)
```

```python
# 배열끼리 더하기
list_var2 = [5,6,7,8]
arr2 = np.array([5,6,7,8])

print(list_var1 + list_var2)
print(arr1 + arr2)
```

```python
# 배열끼리 곱하기
print(arr1 * arr2)
print(list_var1 * list_var2)
```

```python
# 초깃값 있는 1차원 배열의 생성
print(np.zeros(5))
print(np.ones(5))
```

## 시리즈

```python
# 시리즈 연습1
import pandas as pd
# s = pd.Series([10, 4, 3, 26, 12, 9])
s = pd.Series(np.array([10, 4, 3, 26, 12, 9]))
print(type(s))
print(s)
```

```python
# 시리즈 연습2
s.index = ['a', 'b', 'c', 'd', 'e', 'f']
s.name = 'Test'
print(s)
```

```python
# 시리즈 연습3
print(s.loc['c'])
print(s.loc['b':'d'])
print(s.iloc[1:4])
```

```python
# 시리즈 연습4
t = s.iloc[:3]
t.loc['a'] = 100
print(s)
```

```python
# 시리즈 연습5
s['a'] = 10
t = s.iloc[:3].copy()
t['a'] = 100
print(s) 
```

## 선 그래프

```python
# 선 그래프 연습1
import matplotlib.pyplot as plt
import numpy as np

# 데이터 준비
x = np.linspace(-3, 3, 60)  # X축 데이터
y = np.exp(x)  # Y축 데이터

# 그래프 그리기
plt.figure(figsize=(6, 4)) # 크기 설정
plt.plot(x, y)

# 그래프에 제목, 라벨, 그리드 추가
plt.title("Simple Line Plot", fontsize=14)
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(alpha=0.3)

# 그래프 표시
# plt.savefig('da_01.png', dpi=300)
plt.show()
```

```python
# 선 그래프 연습2
x = np.linspace(0.01, 2, 100)
y1 = np.exp(x)
y2 = np.log(x)

plt.figure(figsize=(6, 4))

plt.plot(x, y1, linewidth=1, label='Exp')
plt.plot(x, y2, linewidth=1, label='Log')
plt.axhline(y=0, color='black', linewidth=1)

plt.title("Simple Line Plots", fontsize=14)
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(alpha=0.3)
plt.legend() # 범례 표시
# plt.savefig('plt_02.png', dpi=300)
plt.show()
```

```python
# 선 그래프 연습3
fig, ax = plt.subplots(figsize=(6, 4))

ax.plot(x, y1, linewidth=1, label='Exp')
ax.plot(x, y2, linewidth=1, label='Log')
ax.axhline(y=0, color='black', linewidth=1)
                       
ax.set_title("Simple Line Plots", fontsize=14)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.grid(alpha=0.3)
ax.legend()
plt.show()
```
